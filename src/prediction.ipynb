```python
#importing  libraries
import pandas as pd
import joblib
```

```python
#importing the trained model
model = joblib.load("../model/rf_best_pipeline_log.pkl")
```

```python
#function to predict insurance charges
def predict_insurance_charges(
    model,
    age: float,
    sex: str,
    bmi: float,
    children: int,
    smoker: str,
    region: str
) -> float:

    new_row = pd.DataFrame([{
        "age": age,
        "sex": sex,            
        "bmi": bmi,
        "children": children,
        "smoker": smoker,      
        "region": region       
    }])

    return float(model.predict(new_row)[0])

```

```python
pred = predict_insurance_charges(
    model,
    age=35,
    sex="male",
    bmi=29.5,
    children=2,
    smoker="no",
    region="southeast"
)

print("Predicted Medical Insurance cost", round(pred, 2))
```

```python
#import libraries
import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import mean_absolute_error,root_mean_squared_error, r2_score
```

```python
#import save   best model
rf_best_pipeline_log = joblib.load("../model/rf_best_pipeline_log.pkl")

#x_train, y_train, x_test, y_test
x_train = joblib.load("../model/x_train.pkl")
y_train = joblib.load("../model/y_train.pkl")
x_test = joblib.load("../model/x_test.pkl")
y_test = joblib.load("../model/y_test.pkl")
```

Final Evaluation

```python
#evaluate best log model on train data
train_rf_best_predictions = rf_best_pipeline_log.predict(x_train)

train_rf_best_rmse = root_mean_squared_error(y_train, train_rf_best_predictions)
train_rf_best_mae = mean_absolute_error(y_train, train_rf_best_predictions)
train_rf_best_r2 = r2_score(y_train, train_rf_best_predictions)

print(f"Train MAE: {train_rf_best_mae:.3f}")
print(f"Train RMSE: {train_rf_best_rmse:.3f}")
print(f"Train R2: {train_rf_best_r2:.3f}")
```

```python
#evaluate best log model on test data
test_rf_best_predictions = rf_best_pipeline_log.predict(x_test)

test_rf_best_rmse = root_mean_squared_error(y_test, test_rf_best_predictions)
test_rf_best_mae = mean_absolute_error(y_test, test_rf_best_predictions)
test_rf_best_r2 = r2_score(y_test, test_rf_best_predictions)

print(f"Test MAE: {test_rf_best_mae:.3f}")
print(f"Test RMSE: {test_rf_best_rmse:.3f}")
print(f"Test R2: {test_rf_best_r2:.3f}")
```

```python
#residual plot
residuals = y_test- test_rf_best_predictions

plt.figure(figsize=(10, 6))
sns.scatterplot(x=test_rf_best_predictions, y=residuals)
plt.axhline(0, color='red', linestyle='--') 
plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```

```python
# distribution of residuals
plt.figure(figsize=(6, 4))
sns.histplot(residuals, kde=True, bins=50)
plt.title("Distribution of Residuals")
plt.xlabel("Residuals")
plt.ylabel("Frequency")
plt.show()
```
